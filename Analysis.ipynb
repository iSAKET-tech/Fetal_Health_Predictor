```python
x= df1.iloc[:,:-1]
y= df1.iloc[:,-1]

from sklearn.model_selection import train_test_split
x_train,  x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=1)

```

```python
x_test
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
```

```python
pipe = Pipeline(steps=[
   ('step1' , LogisticRegression(solver= 'lbfgs', max_iter= 1000))
])
```

```python
pipe2 = Pipeline(steps=[
    ('step1', RandomForestClassifier(
        n_estimators= 200,
        random_state= 42
    ) )
])
```

```python
pipe.fit(x_train, y_train)
```

```python
pipe2.fit(x_train, y_train)
```

```python
y_pred =pipe.predict(x_test)
```

```python
y_pred2 = pipe2.predict(x_test)
```

```python
y_pred2
```

```python
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, classification_report

import matplotlib.pyplot as plt
import seaborn as sns
f1_score( y_test, y_pred, average='macro')
```

```python
accuracy_score(y_test, y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

sns.heatmap(
    cm,
    fmt='d',
    xticklabels=['1', '2', '3'],
    yticklabels=['1', '2', '3']
)

plt.xlabel('predicted')
plt.ylabel('actual')
plt.title('confusion_matrix- logistic regression')
plt.show()
```

```python
# RandomFroestClassifier Hyperparameter tuning

from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint
```

```python
param_dist = {
    'step1__n_estimators': randint(100, 500),
    'step1__max_depth': [None, 10, 15, 20, 25, 30],
    'step1__min_samples_split': randint(2, 10),
    'step1__min_samples_leaf': randint(1, 5),
    'step1__max_features': ['sqrt', 'log2', None]
}
```

```python
random_search = RandomizedSearchCV(
    estimator= pipe2,
    param_distributions= param_dist,
    n_iter=30,
    scoring='f1_macro',
    cv=5,
    random_state=42,
    n_jobs=1,
    verbose=1
)
```

```python
random_search.fit(x_train, y_train)
```

```python
print(random_search.best_params_)
```

```python
print(random_search.best_score_)
```

```python
best_model = random_search.best_estimator_

y_pred_tuned = best_model.predict(x_test)
```

```python
y_pred_tuned 
```

```python
f1_score(y_test, y_pred_tuned, average='macro' )
```
